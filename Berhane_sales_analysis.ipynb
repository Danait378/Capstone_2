# EmporiUm Sales Territory Analysis

For this capstone project, I am analyzing in-store sales for two assigned sales territories: Maine and Maryland.
The goal is to compare monthly revenue, store performance, top customers, and product category trends.  
I will use Python, Pandas, and Matplotlib to clean, merge, analyze, and visualize  the sales data.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
customers = pd.read_csv("customer_list.csv", sep="|")

customers.info()
customers.head()
```

```python
stores = pd.read_csv("StoreDetail.csv")

stores.info()
stores.head()
```

```python
stores = pd.read_csv("StoreDetail.csv")

stores.info()
stores.head()
```

```python
sales = pd.read_csv("StoreSales (1).csv")

sales.info()
sales.head()
```

```python
products = pd.read_csv("Products.csv")

products.info()
products.head()
```

```python
categories = pd.read_csv("ProductCategories.csv")

categories.info()
categories.head()
```

# core Marketing Analysis

## Territory Managers and Store Locations

This shows the territory managers, store IDs, and cities for my assigned territories: Maryland and Maine.

```python
maryland_stores = stores[stores["State"] == "maryland"]
maine_stores = stores[stores["State"]== "maine"]
```

## Monthly Revenue by Territory

 Monthly in-store sales revenue for Maryland and Maine.

```python
assigned_stores = stores[
    (stores["State"] == "Maryland") |
    (stores["State"] == "Maine")
]

assigned_stores
```

```python
sales_stores = sales.merge(assigned_stores)

sales_stores["Month"] = pd.to_datetime(
    sales_stores["Transaction Date"]
).dt.month

monthly_revenue = sales_stores.groupby(
    ["State", "Month"]
)["Sale Amount"].sum()

monthly_revenue
```

## Store sales Performance Ranking

```python
store_ranking = sales_stores.groupby(
    ["State", "Store ID"]
)["Sale Amount"].sum().sort_values(ascending=False)

store_ranking
```

## Top Customers by Territory



```python
top_customers = sales_stores.groupby(
    ["State", "RewardsID"]
)["Sale Amount"].sum().reset_index()

top_customers = top_customers.sort_values(
    ["State", "Sale Amount"],
    ascending=[True, False]
)

top_customers
```

## Product Category Analysis



```python
sales_products = sales_stores.merge(products)

sales_products = sales_products.merge(categories)

category_monthly = sales_products.groupby(
    ["State", "Month", "Category"]
).agg({
    "Sale Amount": ["count", "sum"]
})

category_monthly
```

## Chart 1: Monthly Revenue by Territory

This chart shows the monthly sales trends for Maryland and Maine.

```python
assigned_stores = stores[
    (stores["State"] == "Maryland") |
    (stores["State"] == "Maine")
]

sales_stores = sales.merge(assigned_stores, on="Store ID")
```

```python
sales_stores["Transaction Date"] = pd.to_datetime(sales_stores["Transaction Date"])

sales_stores["Month"] = sales_stores["Transaction Date"].dt.month

monthly_revenue = sales_stores.groupby(
    ["State", "Month"]
)["Sale Amount"].sum()

monthly_revenue
```

```python
monthly_revenue = sales_stores.groupby(
    ["State", "Month"]
)["Sale Amount"].sum().reset_index()

monthly_revenue
```

```python
maryland = monthly_revenue[monthly_revenue["State"] == "Maryland"]
maine = monthly_revenue[monthly_revenue["State"] == "Maine"]

plt.plot(maryland["Month"], maryland["Sale Amount"], label="Maryland")
plt.plot(maine["Month"], maine["Sale Amount"], label="Maine")

plt.title("Monthly Revenue by Territory")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.legend()
plt.show()
```
